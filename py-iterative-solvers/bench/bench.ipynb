```python
import numpy as np
import scipy.sparse as sp
```

```python
n = 1024
h = 1.0 / n
a = [2.0 / (h * h)] * (n - 1)
b = [-1.0 / (h * h)] * (n - 2)
mat = np.diag(a) + np.diag(b, k=1) + np.diag(b, k=-1)
rhs = np.array([np.pi**2 * np.sin(i * h * np.pi) for i in range(1, n)])
mat = sp.csr_array(mat)
```

```python
from cg import cg
cg(mat, rhs)
%timeit cg(mat, rhs)
```
